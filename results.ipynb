```python
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt

#Capital 

small_mamba =  [0.03,0.03, 0.01, 0.02, 0.02, 0.03, 0.03, 0.04, 0.04]#, 0.04]
big_mamba = [0.03, 0.06, 0.02, 0.05, 0.06, 0.07, 0.05, 0.06, 0.05]#, 0.06]

small_pythia = [0.04, 0.05, 0.03, 0.04, 0.04, 0.04, 0.06, 0.03, 0.06]#, 0.07]
big_pythia = [0.03, 0.04, 0.03, 0.04, 0.04, 0.04, 0.06, 0.03, 0.03]#, 0.06]
x_values = [1, 2, 4, 5, 8, 16, 32, 64, 128]#, 256]  # Original x-values



# Set seaborn style
sns.set(style="whitegrid")

# Create the plot using Seaborn
fig, ax1 = plt.subplots(figsize=(6, 4))

# Plot each model with connecting lines
sns.lineplot(ax=ax1, x=x_values, y=small_mamba, label='130m Mamba', marker='o', color='b', linestyle=":")
sns.lineplot(ax=ax1, x=x_values, y=big_mamba, label='2.8b Mamba', marker='o', color='g', linestyle=":")
sns.lineplot(ax=ax1, x=x_values, y=small_pythia, label='160m Pythia', marker='s', color='r', linestyle=":")
sns.lineplot(ax=ax1, x=x_values, y=big_pythia, label='2.8b Pythia', marker='s', color='m', linestyle=":")




# Set logarithmic scale on x-axis using ax1
ax1.set_xscale('log')

ax1.set_ylim(0.0, 0.25)

# Limit y-axis from 0.0 to 0.5

# Labels and title
ax1.set_xlabel('# In-Context Examples')
ax1.set_ylabel('Accuracy')

# Custom x-tick labels (powers of 10)

# Show grid and legend
ax1.grid(True)
ax1.legend()
plt.title("Name to city mapping: In-Context Length vs. Accuracy")

# Display the plot
plt.show()
```
